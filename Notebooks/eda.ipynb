```python
import pandas as pd
import numpy as np

df = pd.read_csv(r"C:\Users\olale\Documents\Lekan\vehicles_us.csv.csv")

df['date_posted'] = pd.to_datetime(df['date_posted'])

df = df[df['price'] >= 100]
df = df[df['odometer'] <= 400_000]

df['model_year'] = df['model_year'].fillna(df['model_year'].median())


df['cylinders'] = df.groupby('type')['cylinders'].transform(
    lambda x: x.fillna(x.median())
)


df['odometer'] = df['odometer'].fillna(df['odometer'].median())


df['is_4wd'] = df['is_4wd'].fillna(0)


df['paint_color'] = df['paint_color'].fillna('unknown')


df['vehicle_age'] = df['date_posted'].dt.year - df['model_year']


df['price_per_mile'] = df['price'] / df['odometer']


df['log_price'] = np.log1p(df['price'])


df['posting_month'] = df['date_posted'].dt.month


print(df.shape)
df.head()



import plotly.express as px

fig = px.histogram(
    df,
    x="price",
    color="condition",
    nbins=50,
    title="Vehicle Price Distribution by Condition",
    labels={"price": "Price ($)", "condition": "Condition"},
    opacity=0.7
)

fig.update_layout(
    xaxis_range=[0, 60000],
    bargap=0.1
)

fig.show()


fig = px.histogram(
    df,
    x="odometer",
    color="type",
    nbins=60,
    title="Mileage Distribution by Vehicle Type",
    labels={"odometer": "Mileage"},
)

fig.update_layout(
    xaxis_range=[0, 300000]
)

fig.show()

fig = px.scatter(
    df,
    x="vehicle_age",
    y="price",
    color="type",
    title="Vehicle Price vs Age by Vehicle Type",
    labels={"vehicle_age": "Vehicle Age (Years)", "price": "Price ($)"},
    opacity=0.6
)

fig.update_layout(
    yaxis_range=[0, 60000]
)

fig.show()




```
